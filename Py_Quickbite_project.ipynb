__Importing Libraries__

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

__Loading the Datasets__

```python
customers=pd.read_csv("OneDrive/Documents/Project_QuickBite/Project_QuickBite/customers.csv")
menu_items=pd.read_csv("OneDrive/Documents/Project_QuickBite/Project_QuickBite/menu_items.csv")
orders=pd.read_csv("OneDrive/Documents/Project_QuickBite/Project_QuickBite/orders.csv")
payments=pd.read_csv("OneDrive/Documents/Project_QuickBite/Project_QuickBite/payments.csv")
restaurants=pd.read_csv("OneDrive/Documents/Project_QuickBite/Project_QuickBite/restaurants.csv")
```

__Merging Datasets__

```python
customer_orders = customers.merge(orders, on='customer_id')
```

```python
order_restaurants = orders.merge(restaurants, on='restaurant_id')
```

```python
order_payments = orders.merge(payments, on='order_id')
```

```python
restaurant_menu = restaurants.merge(menu_items, on='restaurant_id')
```

```python
#Checking the null values in merged datasets
print(customer_orders.isnull().sum())

print(order_restaurants.isnull().sum())

print(order_payments.isnull().sum()) 

print(restaurant_menu.isnull().sum())
```

__Combined Analysis__

```python
#top 10 customer of highest total spent.
customer_spending = order_payments.groupby('customer_id')['order_amount'].sum()
customer_spending.sort_values(ascending=False).head(10)
```

```python
customer_spending.plot(kind='bar')
plt.xlabel('customers_id')
plt.ylabel('order_amount')
plt.title('highest customers_spendings')
plt.show()
```

```python
#Find which payment method accounts having highest total payment amount.
payment_method=order_payments.groupby('payment_method')['order_amount'].sum().sort_values(ascending=False)
payment_method
```

```python
payment_method.plot(kind='pie',autopct='%1.1f%%')
plt.title('total amount by payment_method')
plt.ylabel('')                   
plt.show()
```

```python
#top 10 restaurants having highest revenue
orders.groupby('restaurant_id')['order_amount'].sum().sort_values(ascending=False).head(10)
```

```python
#finding how many orders were placed by each customer segment.
customer_orders_by_segment=customer_orders.groupby('customer_segment')['order_id'].count()
customer_orders_by_segment
```

```python
customer_orders_by_segment.plot(kind='bar')
plt.xlabel('customer segment')
plt.ylabel('order_id')
plt.title('customer orders by segment')
plt.show()
```

```python
#comparing restaurant revenue and order count
restaurant_analysis = orders.groupby('restaurant_id').agg(
    total_revenue=('order_amount', 'sum'),
    total_orders=('order_id', 'count')
)
restaurant_analysis
```

```python
restaurant_analysis.plot(kind='bar',secondary_y='total_orders')
plt.xlabel('Restaurant id')
plt.ylabel('total Revenue and total Count')
plt.title(' total Revenue vs total Order')
plt.show()
```

```python
#converting order_date column to date format
orders['orders_date']=pd.to_datetime(orders['order_date'])
```

```python
#how many unique customers placed orders each month.
monthly_orders=orders.groupby([orders['orders_date'].dt.year,orders['orders_date'].dt.month]).size()
monthly_orders
```

```python
monthly_orders.plot(kind='line',marker='o')
plt.xlabel('month')
plt.ylabel('num of orders')
plt.title('monthly orders count')
plt.show()
```

```python
#Total Revenue month wise
order_restaurants['orders_date']=pd.to_datetime(order_restaurants['order_date'])
monthly_revenue=order_restaurants.groupby(order_restaurants['orders_date'].dt.to_period('M'))['order_amount'].sum()
monthly_revenue
```

```python
monthly_revenue.plot(kind='area')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.title('Monthly Revenue Trend')
plt.show()
```

```python
This chart shows total revenue trend of each month
June month has the highest revenue
Sep month has the lowest revenue
```

```python
#Distribution of order amount
order_restaurants['order_amount'].plot(kind='hist', bins=10)
plt.xlabel('Order Amount')
plt.ylabel('Number of Orders')
plt.title('Distribution of Order Amounts')
plt.show()
```

```python
This chart shows how many orders fall within different price ranges.
```
